### Question 1 — Line Plot  `Matplotlib`

Calculate the **average marks** of all students in each of the four tests (`Test_1` to `Test_4`). Create a **line plot** to show how the class average changes from test to test. Add a title and axis labels.

```python
# ✏️ Your code here
# avg1=float(np.mean(df["Test_1"]))
# avg2=float(np.mean(df["Test_2"]))
# avg3=float(np.mean(df["Test_3"]))
# avg4=float(np.mean(df["Test_4"]))
averages=np.mean(df[["Test_1","Test_2","Test_3","Test_4"]],axis=0)
plt.plot(averages,marker="o")
plt.title("Average marks across tests")
plt.xlabel("Tests")
plt.ylabel("Average")
plt.show()
```

### Question 2 — Bar Chart  `Matplotlib`

Create a **bar chart** comparing the final `Marks` of the **first five students** in the dataset.

### Question 3 — Histogram  `Matplotlib`

Create a **histogram** of the final `Marks` of all 30 students. Change the **number of bins** (for example 3, 5 and 10) and write what difference you observe.

```python
for b in [3, 5, 10]:

    plt.hist(df["Marks"], bins=b, edgecolor="black")

    plt.title(f"Distribution of Marks - {b} Bins")
    plt.xlabel("Marks")
    plt.ylabel("Frequency")

    plt.show()
    print("\n")
```

🔎 Observation: *Mark distribution is high in the 80 to 100 range*

# 🟢 Part B: Seaborn Questions

### Question 6 — Count Plot  `Seaborn`

Using **Seaborn**, create a **count plot** showing the number of students in each `Class`.

```python
sns.countplot(data=df, x="Class")

plt.title("Number of Students in Each Class")
plt.xlabel("Class")
plt.ylabel("Number of Students")

plt.show()
```

### Question 7 — Box Plot  `Seaborn`

Using **Seaborn**, create a **box plot** to compare the final `Marks` of **Class A** and **Class B**. Which class has the higher median?

```python
sns.boxplot(data=df, x="Class", y="Marks")

plt.title("Comparison of Final Marks Between Classes")
plt.xlabel("Class")
plt.ylabel("Final Marks")

plt.show()

#class A has higher median
```

### Question 8 — Heatmap  `Seaborn`

Create a **heatmap** of the correlation between `Study_Hours`, `Attendance`, `Sleep_Hours` and `Marks`. Show the values on the heatmap. Which column has the strongest relationship with `Marks`?

```python
# ✏️ Your code here

numericals=df[["Study_Hours", "Attendance", "Sleep_Hours", "Marks"]]

correlation=numericals.corr()

sns.heatmap(data=correlation,annot=True,cmap="coolwarm")

plt.title("Correlation Heatmap")
plt.xlabel("Features")
plt.ylabel("Features")

plt.show()

# Study_Hours has the strongest relationship with Marks.
```
